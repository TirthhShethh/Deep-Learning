#Sentiment Analysis on Movie Reviews

```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np
```

```python
# 1. Load the Data
# num_words=10000: We only keep the 10,000 most common words. Rare words are ignored.
print("Loading data...")
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.imdb.load_data(num_words=10000)
```

```python
# 2. Preprocessing (Padding)
# We force every review to be exactly 256 words long.
x_train = tf.keras.preprocessing.sequence.pad_sequences(x_train, maxlen=256)
x_test = tf.keras.preprocessing.sequence.pad_sequences(x_test, maxlen=256)
```

```python
print(f"Training data shape: {x_train.shape}")
print(f"A single review (in numbers): {x_train[0][:10]} ...")
```

```python
# 3. Build the RNN Model
model = tf.keras.Sequential([
    # Layer 1: Embedding
    # Turns positive integers (indexes) into dense vectors of fixed size.
    # Imagine a 3D map where "Good" and "Great" are close, but "Bad" is far away.
    layers.Embedding(input_dim=10000, output_dim=32),
    # Layer 2: SimpleRNN
    # Reads the input sequence step-by-step.
    # 32 units = The size of its "memory".
    layers.LSTM(32),

    # Layer 3: Dense (Output)
    # Single neuron. If output > 0.5, it's Positive. If < 0.5, it's Negative.
    layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy', # Used for Yes/No classification
              metrics=['accuracy'])
```

```python
# 4. Train
print("Training model...")
history = model.fit(x_train, y_train, epochs=5, batch_size=128, validation_split=0.2)
```

```python
# 5. Evaluate
results = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {results[1]*100:.2f}%")
```
